```python
df = pd.read_csv('../data/jobs.csv')

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.loc[2, 'Salary_LPA'] = None
df.loc[5, 'Skills'] = None
df.loc[7, 'Location'] = None

df
```

```python
df['Salary_LPA'].fillna(df['Salary_LPA'].mean(), inplace=True)
```

```python
df['Salary_LPA'] = df['Salary_LPA'].fillna(df['Salary_LPA'].mean())
```

```python
df['Skills'] = df['Skills'].fillna("Not Specified")
df['Location'] = df['Location'].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df['Location'].value_counts()
```

```python
df['Salary_LPA'].mean()
```

```python
df['Experience_Level'].value_counts()
```

```python
skills_series = df['Skills'].str.split(', ').explode()
skills_series.value_counts()
```

```python
import matplotlib.pyplot as plt

df['Location'].value_counts().plot(kind='bar')
plt.title("Jobs by Location")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

```python
skills_series = df['Skills'].str.split(', ').explode()

plt.figure()
skills_series.value_counts().plot(kind='bar')
plt.title("Top Skills in Demand")
plt.xlabel("Skills")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
sns.countplot(x='Location', data=df)
plt.title("Job Distribution by Location")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
sns.histplot(df['Salary_LPA'], kde=True)
plt.title("Salary Distribution (LPA)")
plt.show()
```

```python
plt.figure()
sns.boxplot(x='Experience_Level', y='Salary_LPA', data=df)
plt.title("Salary vs Experience Level")
plt.show()
```
